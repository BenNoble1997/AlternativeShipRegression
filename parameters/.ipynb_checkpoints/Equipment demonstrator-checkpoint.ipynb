### LoggerMixin

A mixin is a special kind of class that's designed to add functionality to other classes through inheritance. It's like a set of pre-built features that you can plug into any class that needs them. The 'LoggerMixin' adds logging capabilities to any class that inherits from it.

They are generally reccomended against (for reasons i can only parrot) but this is the exception to the rule as it means you dont have to set up logging every time.

### Fuel class
This is a dataclass which stores the fuel properties, which are then called by the equipment classes.
Parameters stored:
* name of fuel
* fuel type (e.g. hydrogen, methanol, methane, diesel)
* LHV - KJ/kg
* rho - Density
* EF - kg of CO2 per kg of fuel
* p_fuel - fuel price in dollars per tonne
* s_price - storage cost of the fuel (someone else used it once so i kept it just in case)

```python
class Fuel(LoggerMixin):
    """
    A class representing fuel properties used in ship energy systems.

    """
    def __init__(self,
                 name:str,
                 fuel_type:FuelType,
                 LHV:int,
                 rho:float,
                 EF:float,
                 p_fuel:float=0,
                 s_price:float =0
                 ):
        """
        Initialize a Fuel object with specified attributes.
        Handles immediate type checking and simple range validation

        :param name: Name of the fuel.
        :type name: str

        :param formula: Chemical formula of the fuel.
        :type formula: str

        :param LHV: Lower calorific value (LHV) of the fuel in kilojoules per kilogram (kJ/kg).
        :type LHV: float

        :param rho: Density of the fuel in kilograms per cubic meter (kg/m**3).
        :type rho: float

        :param p_fuel: Price of the fuel in dollars per ton ($/ton), as of 2020 and 2030 based on Baldi et al.
        :type p_fuel: float

        :param s_price: Cost of fuel storage (USD/kg)
        :type s_price: float

        :param EF: Emission factor of the fuel in tons CO2 equivalent per ton of fuel, based on IMO fourth GHG study, 2020.
        :type EF: float
        """
        self.name = name
        self.fuel_type = fuel_type
        self.LHV = LHV
        self.rho = rho
        self.p_fuel = p_fuel
        self.s_price = s_price
        self.EF = EF

        if not isinstance(name,str):
            raise TypeError('Name must be a string.')

        cleaned_name = name.strip().title()
        if not cleaned_name:
            raise ValueError('Name cannot be empty.')

        try:
            LHV = float(LHV)
            rho = float(rho)
            p_fuel = float(p_fuel)
            s_price = float(s_price)
            EF = float(EF)
        except (TypeError, ValueError) as e:
            raise TypeError(f"All numerical values must be convertible to float: {str(e)}")

    def __post_init__(self):
        """
        Perform complex validation after initialization.
        This handles cross-field validation and business rules.
        """
        # Validate physical properties are within reasonable bounds
        if not PHYSICAL_LIMITS['min_LHV'] <= self.LHV <= PHYSICAL_LIMITS['max_LHV']:
            raise ValueError(
                f"LHV {self.LHV} kJ/kg is outside physical limits "
                f"({PHYSICAL_LIMITS['min_LHV']}, {PHYSICAL_LIMITS['max_LHV']})"
            )

        if not PHYSICAL_LIMITS['min_density'] <= self.rho <= PHYSICAL_LIMITS['max_density']:
            raise ValueError(
                f"Density {self.rho} kg/m³ is outside reasonable limits "
                f"({PHYSICAL_LIMITS['min_density']}, {PHYSICAL_LIMITS['max_density']})"
            )

        # Validate prices are non-negative
        if self.p_fuel < 0 or self.s_price < 0:
            raise ValueError("Prices cannot be negative")

        # Validate emission factor is positive
        if self.EF <= 0:
            raise ValueError("Emission factor must be positive")

        # Validate chemical formula format and composition
        if self.formula:
            # Check if formula matches any standard fuel type
            try:
                fuel_type = FuelType(self.formula)
            except ValueError:
                # If not a standard fuel, validate formula format
                if not re.match(r'^[a-z0-9]+$', self.formula):
                    raise ValueError(
                        "Formula can only contain letters and numbers"
                    )

        # Cross-validate energy density
        energy_density = self.LHV * self.rho
        if energy_density > 120_000_000:  # Maximum theoretical energy density in kJ/m³
            raise ValueError(
                f"Calculated energy density {energy_density} kJ/m³ exceeds "
                "physical limits for chemical fuels"
            )

    def __repr__(self) -> str:
        """
        Return a string representation of the Fuel object.

        Returns:
            str: String representation including the name and key attributes of the Fuel object.
        """
        return f"Fuel(name='{self.name}', LHV={self.LHV}, rho={self.rho}, price={self.p_fuel}, EF={self.EF})"

    @property
    def cEDV(self) -> float:
        """
        Calculate and return the contained volumetric energy density of the fuel.

        :return: Contained volumetric energy density of the fuel in gigajoules per cubic meter (GJ/m**3).
        :rtype: float
        """
        return (self.LHV * self.rho) / 1000


    @staticmethod
    def fuels_to_dataframe(fuels: list) -> pd.DataFrame:
        """
        Convert a list of Fuel objects into a pandas DataFrame.

        :param fuels: List of Fuel objects.
        :type fuels: list
        :return: DataFrame containing the attributes of each Fuel object.
        :rtype: pd.DataFrame
        """
        data = {
            'Name': [fuel.name for fuel in fuels],
            'Formula': [fuel.formula for fuel in fuels],
            'LHV (kJ/kg)': [fuel.LHV for fuel in fuels],
            'Density (kg/m^3)': [fuel.rho for fuel in fuels],
            'Fuel Price ($/ton)': [fuel.p_fuel for fuel in fuels],
            'Storage Price (USD/kg)': [fuel.s_price for fuel in fuels],
            'Emission Factor (tons CO2/ton)': [fuel.EF for fuel in fuels],
            'Contained Volumetric Energy Density (GJ/m^3)': [fuel.cEDV for fuel in fuels]
        }

        return pd.DataFrame(data)

```

### Equipment Configurations

Configuration data such as parameters, limits and efficiencty curves are kept separate from the class instance that uses them. Enables them to be more easily varied for different typese of engine etc.
Also for if better info appears down the line (e.g. I am currently using the SFOC coefficients from that paper)

Advantages:
* Parameters can be easily modified without touching class logic
* Defaults provides fallback for undefined parameters (And I can add a log message saying that it is falling back to the default)

```python
@dataclass
class ICEConfig:
    """
    Configuration parameters for Internal Combustion Engines (ICE),
    structured to provide multiple approaches for efficiency modeling.
    """
    # Operating limits as defined in your get_operating_limits method
    operating_limits: Dict[str, Dict[str, float]] = field(default_factory=lambda: {
        'DEFAULT': {
            'min_load': 0.2,  # 20% of rated power
            'max_load': 1.0,  # 100% of rated power
            'max_ramp_rate': 0.1  # 10% per minute
        },
        'BASIC':{
            'min_load': 0.2,  # 20% of rated power
            'max_load': 1.0,  # 100% of rated power
            'max_ramp_rate': 0.1  # 10% per minute
        }
    })
    # Base SFOC for each engine type with default fuel (usually diesel)
    base_sfoc: Dict[str, float] = field(default_factory=lambda: {
        'DEFAULT': 180.0,  # g/kWh for diesel
        'BASIC': 180.0,
    })

    # SFOC correction factors for different fuels
    # These multipliers adjust the base SFOC for different fuel types
    fuel_sfoc_factors: Dict[str, Dict[FuelType, float]] = field(default_factory=lambda: {
        'DEFAULT': {
            FuelType.DIESEL: 1.0,  # Reference fuel (no adjustment)
            FuelType.HFO: 1.05,  # HFO typically has ~5% higher consumption
            FuelType.METHANE: 0.85,  # Natural gas often has lower mass consumption
            FuelType.HYDROGEN: 0.3,  # Much lower mass consumption due to high energy content
        },
        # Add other engine types as needed
    })

    # SFOC calculation coefficients (shape of the curve)
    sfoc_coefficients: Dict[str, Dict[str, float]] = field(default_factory=lambda: {
        'DEFAULT': {
            'a': 0.455,  # Coefficient for load_point^2
            'b': -0.71,  # Coefficient for load_point
            'c': 1.28  # Constant term
        },
        'BASIC': {
            'a': 0.0,
            'b': 0.0,
            'c': 2.5
        }
    })
    # SFOC calculation coefficients from your fuel_consumption_at_load method
    sfoc_coefficients: Dict[str, Dict[str, float]] = field(default_factory=lambda: {
        'DEFAULT':{
            'a': 0.455,  # Coefficient for load_point^2
            'b': -0.71,  # Coefficient for load_point
            'c': 1.28  # Constant term
        },
        'BASIC':{
            'a': 0.0,    # No quadratic component
            'b': 0.0,    # No linear component
            'c': 2.5     # Constant value of 2.5 gives efficiency of 0.4
        }
    })

    # Direct efficiency curves for different engine types
    # Keys are load points (0.0 to 1.0), values are efficiency (0.0 to 1.0)
    efficiency_curves: Dict[str, Dict[float, float]] = field(default_factory=lambda: {
        'BASIC': {
            0.0: 0.0,  # No efficiency at zero load
            0.1: 0.4,  # Very low efficiency at 10% load
            0.2: 0.4,  # Efficiency starts improving
            0.3: 0.4,
            0.4: 0.4,
            0.5: 0.4,
            0.6: 0.4,
            0.7: 0.4,
            0.8: 0.4,  # Peak efficiency around 80% load
            0.9: 0.4,
            1.0: 0.4  # Slight drop at 100% load
        },
        'DEFAULT': None,  # No default curve, will fall back to SFOC calculation
        'EXAMPLE': {
            0.0: 0.0,  # No efficiency at zero load
            0.1: 0.15,  # Very low efficiency at 10% load
            0.2: 0.28,  # Efficiency starts improving
            0.3: 0.36,
            0.4: 0.42,
            0.5: 0.47,
            0.6: 0.50,
            0.7: 0.52,
            0.8: 0.535,  # Peak efficiency around 80% load
            0.9: 0.53,
            1.0: 0.515  # Slight drop at 100% load
    })

    # Lookup table for engine fuel compatibility
    fuel_compatibility: Dict[str, List[FuelType]] = field(default_factory=lambda: {
        'DEFAULT': [FuelType.DIESEL, FuelType.HFO, FuelType.METHANE],
        'MEDIUM_SPEED_DIESEL': [FuelType.DIESEL, FuelType.HFO],
        'DUAL_FUEL_ICE': [FuelType.DIESEL, FuelType.METHANE, FuelType.HYDROGEN],
        'GAS_TURBINE': [ FuelType.DIESEL, FuelType.METHANE]})

    def get_sfoc_for_fuel(self, engine_type: str, fuel_type: FuelType) -> float:
        """
        Get the base SFOC value for a specific engine-fuel combination.

        Args:
            engine_type: The type of engine
            fuel_type: The type of fuel

        Returns:
            Base SFOC value in g/kWh
        """
        # Default to 'DEFAULT' if engine type not found
        if engine_type not in self.base_sfoc:
            engine_type = 'DEFAULT'

        # Get base SFOC for this engine
        base = self.base_sfoc[engine_type]

        # Get fuel factor (default to 1.0 if not specified)
        factor = 1.0
        if engine_type in self.fuel_sfoc_factors:
            factor = self.fuel_sfoc_factors[engine_type].get(fuel_type, 1.0)
        elif 'DEFAULT' in self.fuel_sfoc_factors:
            # Fall back to DEFAULT engine type factors
            factor = self.fuel_sfoc_factors['DEFAULT'].get(fuel_type, 1.0)

        return base * factor
```

### Abstract class
The concept here essentially works as a contract for any energy covnerter class I make.
Any energy converter class must have these three methods minimum.
Enables me to type hint EnergyConverter in my code, which makes it more readable (supposedly)

```python
class EnergyConverter(ABC):
    @abstractmethod
    def get_operating_limits(self) -> Dict[str, float]:
        pass

    @abstractmethod
    def efficiency_at_load(self, load: float) -> float:
        pass

    @abstractmethod
    def fuel_consumption_at_load(self, load: float, fuel: 'Fuel') -> float:
        pass
```

#### Efficiency at load
Two different approaches for calculating efficiency
1) Using interpolation of an efficiency curve (essentially a lookup table)
2) Polynomial approach.
Essentially depends on what you have avaiable. The DT4GS model had one and the other



```python
def efficiency_at_load(self, load: float) -> float:
    load_point = load_point(load)

    if self.efficiency_curve is not None:
        self.logger.info(f"Calculating efficiency at {load_point} using efficiency curve data")
        return np.interp(load_point,
                         list(self.efficiency_curve.keys()),
                         list(self.efficiency_curve.values()))

    elif self.efficiency_coeffs is not None:
        self.logger.info(f"Calculating efficiency at {load_point} using efficiency coefficient data")
        a = self.efficiency_coeffs['a']
        b = self.efficiency_coeffs['b']
        c = self.efficiency_coeffs['c']
        return a * (load_point ** 2) + b * load_point + c
```

```python
class PowerSharingStrategy(Enum):
    """Defines different strategies for sharing power between multiple converters."""
    EQUAL = "equal"  # Split power equally between converters
    PROPORTIONAL = "proportional"  # Split power proportional to rated power
    OPTIMAL_EFFICIENCY = "optimal_efficiency"  # Split power to maximize overall efficiency
    CUSTOM = "custom"  # Use custom ratio provided in configuration

@dataclass
class PowerSharingConfig:
    """Configuration for power sharing between multiple converters."""
    strategy: PowerSharingStrategy = PowerSharingStrategy.PROPORTIONAL
    custom_ratios: Optional[List[float]] = None  # Used with CUSTOM strategy
    min_active_converters: int = 1  # Minimum number of converters that must be active
    load_threshold: float = 0.2  # Minimum load threshold for activating additional converter

    def __post_init__(self):
        """Validate configuration after initialization."""
        if self.strategy == PowerSharingStrategy.CUSTOM and not self.custom_ratios:
            raise ValueError("Custom ratios must be provided when using CUSTOM strategy")
        if self.custom_ratios and abs(sum(self.custom_ratios) - 1.0) > 1e-6:
            raise ValueError("Custom ratios must sum to 1.0")
        if self.load_threshold <= 0 or self.load_threshold >= 1:
            raise ValueError("Load threshold must be between 0 and 1")
```

```python
    def _determine_active_converters(self,
                                     energy_converter:Union[EnergyConverter, List[EnergyConverter]],
                                     total_power: float) -> int:
        """
        Determine how many converters should be active based on power requirement.

        :TODO: need to factor in the fact that this is only really relevant in the event of generators
                power is not shared when its an engien attached to a shaft 

        :param total_power: Power required by the converters
        :type total_power: float

        :return: reccomended number of converters.
        """
        min_active = self.power_sharing_config.min_active_converters
        threshold = self.power_sharing_config.load_threshold

        # Calculate total rated power
        total_rated = sum(conv.rated_power for conv in energy_converter)

        # Calculate load ratio
        load_ratio = total_power / total_rated

        # Determine number of active converters needed
        if load_ratio <= threshold:
            return min_active

        recommended_active = max(
            min_active,
            min(
                len(energy_converter),
                ceil(load_ratio / threshold)
            )
        )

        return recommended_active
```

#### Power distribution calculation

determines the power distribution of the engines being considered. 
Input: An EnergyConverter or a list of energy converters

step 1) number of converters
    Check the input
        - if a single EnergyConverter it is one
        if a list of converters - number of converters is determined
    If the input is one, the total power is 1, otherwise it is then divided as follows
    
step 2) get the strategy from the power sharing config
        determine the number of active converters - want to come back to this

step 3) divide the power as per the strategy

returns a list of power levels that matches up with the list of converters it was supplied with.



```python
    def _calculate_power_distribution(self,
                                      energy_converter:Union[EnergyConverter, List[EnergyConverter]],
                                      total_power: float) -> List[float]:
        """
        Calculate how power should be distributed among multiple converters.

        :param total_power: Total power requirement to be distributed
        :type total_power: float

        :return: List of power allocations for each converter
        :rtype: List[float]

        """
        if isinstance(energy_converter, EnergyConverter):
            num_converters = 1
        elif isinstance(energy_converter, list):
            num_converters = len(energy_converter)
        else:
            raise ValueError("energy_converter must be an instance of EnergyConverter or a list of EnergyConverters")
        if num_converters == 1:
            self.logger.info(f"power covered by one converter")
            return [total_power]

        strategy = self.power_sharing_config.strategy

        # Determine number of active converters needed
        active_converters = self._determine_active_converters(total_power)
        self.logger.info(f"active converters: {active_converters}")

        if strategy == PowerSharingStrategy.EQUAL:
            # Split power equally among active converters
            self.logger.info(f"strategy for splitting main power is splitting power equally")
            power_per_converter = total_power / active_converters
            return [power_per_converter if i < active_converters else 0.0
                    for i in range(num_converters)]

        elif strategy == PowerSharingStrategy.PROPORTIONAL:
            # Split power proportional to rated power
            self.logger.info("Power shared proportionally to rated power of converters")
            total_rated_power = sum(conv.rated_power for conv in energy_converter[:active_converters])
            return [
                (conv.rated_power / total_rated_power * total_power)
                if i < active_converters else 0.0
                for i, conv in enumerate(energy_converter)
            ]

        elif strategy == PowerSharingStrategy.OPTIMAL_EFFICIENCY:
            # Use optimization to find most efficient power distribution
            #This method has not been devised yet
            self.logger.info("Determining power sharing using optimal_efficiency calculator")
            return self._optimize_power_distribution(total_power, active_converters)

        elif strategy == PowerSharingStrategy.CUSTOM:
            # Use provided custom ratios
            ratios = self.power_sharing_config.custom_ratios[:active_converters]
            self.logger.info(f"power sharing using specified ratio: {ratios}")
            normalized_ratios = [r / sum(ratios) for r in ratios]
            return [
                (ratio * total_power if i < active_converters else 0.0)
                for i, ratio in enumerate(normalized_ratios)
            ]

        raise ValueError(f"Unknown power sharing strategy: {strategy}")
```

#### Calculating_system_efficiency

        - The proportion of power that they provide
step 3) Include the efficiencies of other equipment. e.g. shaft motor, shaft generator, reformer etc - _This is not done in the most robust way_


```python
 def calculate_system_efficiency(self,
                                    power_required: float,
                                    energy_converter: Union[EnergyConverter, List[EnergyConverter]],
                                    electricity_converter_required: bool = False,
                                    fuel_reformer_required: bool = False) -> float:
        """
        Calculate overall system efficiency with power sharing for multiple converters.

        This method computes a weighted average efficiency based on each converter's
        contribution to the total power, then applies additional system losses
        from reformers or electricity converters if required.
        """
        self.logger.info(f"Calculating system efficiency for {power_required:.2f} kW")

        # Get power distribution across converters
        power_distribution = self._calculate_power_distribution(
            total_power=power_required,
            energy_converter=energy_converter
        )

        # Ensure energy_converter is a list for consistent processing
        if isinstance(energy_converter, EnergyConverter):
            self.logger.debug("Single converter provided - converting to list for processing")
            energy_converter = [energy_converter]

        # Log the converter count and types
        converter_types = [f"{conv.__class__.__name__}({conv.name})" for conv in energy_converter]
        #The above line looks scary but just gets the name of the converter and the type
        self.logger.info(f"System contains {len(energy_converter)} converters: {', '.join(converter_types)}")

        # Log power distribution
        distribution_summary = ", ".join([f"{converter_types[i]}: {power:.2f} kW"
                                          for i, power in enumerate(power_distribution) if power > 0])
        self.logger.info(f"Power distribution: {distribution_summary}")

       
     # Calculate individual and weighted efficiencies
        total_efficiency_contribution = 0.0
        total_power = 0.0

        converter_efficiencies = []


     #zip creates a tuple of each converter and corresponding power
     ##enumerate indexes each of these tuples which is handy for then going through them
        for i, (converter, power) in enumerate(zip(energy_converter, power_distribution)):
            if power > 0:
                # Get individual converter efficiency at its load point
                load_point = power / converter.rated_power
                converter_efficiency = converter.efficiency_at_load(power)
                converter_efficiencies.append((i, converter.name, power, load_point, converter_efficiency))

                # Calculate weighted contribution
                ##Weighted in terms of the efficiency and the proportion of the power that they are providing
                weighted_contribution = converter_efficiency * power
                total_efficiency_contribution += weighted_contribution
                total_power += power

                self.logger.debug(
                    f"Converter {i} ({converter.name}): "
                    f"Power={power:.2f}kW ({load_point * 100:.1f}% load), "
                    f"Efficiency={converter_efficiency:.4f}, "
                    f"Contribution={weighted_contribution:.2f}"
                )

        # Log all converter efficiencies in a table-like format
        efficiency_table = "\n    ".join([
            f"Converter {i}: {name}, {power:.2f}kW, {load_point * 100:.1f}% load, η={efficiency:.4f}"
            for i, name, power, load_point, efficiency in converter_efficiencies
        ])
        self.logger.info(f"Individual converter efficiencies:\n    {efficiency_table}")

        # Calculate weighted average efficiency
        if total_power > 0:
            avg_efficiency = total_efficiency_contribution / total_power
            self.logger.info(
                f"Base weighted efficiency (before additional factors): {avg_efficiency:.4f} ({avg_efficiency * 100:.2f}%)")
        else:
            avg_efficiency = 0.0
            self.logger.warning("Zero power requirement - returning zero efficiency")
            return 0.0

        # Apply additional efficiency factors
        original_efficiency = avg_efficiency

        # Apply reformer efficiency if required
        if fuel_reformer_required:
            if self.reformer:
                reformer_efficiency = self.reformer.efficiency
                self.logger.info(
                    f"Applying reformer efficiency: {reformer_efficiency:.4f} "
                    f"({self.reformer.__class__.__name__}: {self.reformer.name})"
                )
                avg_efficiency *= reformer_efficiency
            else:
                default_reformer_efficiency = 0.5
                self.logger.warning(
                    f"No reformer configured but fuel_reformer_required=True. "
                    f"Using default efficiency factor: {default_reformer_efficiency}"
                )
                avg_efficiency *= default_reformer_efficiency

        # Apply electricity converter efficiency if required
        if electricity_converter_required:
            if self.electricity_converter:
                converter_efficiency = self.electricity_converter.efficiency
                self.logger.info(
                    f"Applying electricity converter efficiency: {converter_efficiency:.4f} "
                    f"({self.electricity_converter.__class__.__name__}: {self.electricity_converter.name})"
                )
                avg_efficiency *= converter_efficiency
            else:
                default_converter_efficiency = 0.8
                self.logger.warning(
                    f"No electricity converter configured but electricity_converter_required=True. "
                    f"Using default efficiency factor: {default_converter_efficiency}"
                )
                avg_efficiency *= default_converter_efficiency

        # Log efficiency changes
     #original efficiency is the efficiency of the combined converters
        total_losses = original_efficiency - avg_efficiency
        percent_losses = (total_losses / original_efficiency) * 100 if original_efficiency > 0 else 0

        self.logger.info(
            f"Final system efficiency: {avg_efficiency:.4f} ({avg_efficiency * 100:.2f}%) "
            f"[{percent_losses:.2f}% lost in additional conversion steps]"
        )

        return avg_efficiency
```

#### Calculating fuel consumption
This function is similar to the previous one. 

step 1) power distribution

step 2) Calculate corresponding fuel consumption for each converter at specific power level.

step 3) multiply by number of hours to get the mass of fuel used in kilograms


```python
    def calculate_fuel_consumption(self,
                                   power_required: float,
                                    energy_converter:Union[EnergyConverter, List[EnergyConverter]],
                                   fuel: Fuel,
                                   duration_hours: float) -> float:
        """
        Calculate total fuel consumption with power sharing.

        :param power_required: Power required by the converters
        :type power_required: float
        :param energy_converter: Energy converters
        :type energy_converter: List[EnergyConverter]
        :param fuel: Fuel
        :type fuel: Fuel
        :param duration_hours: Duration of fuel consumption
        :type duration_hours: float

        :return: Total fuel consumption
        :rtype: float
        """

        if isinstance(energy_converter, EnergyConverter):
            energy_converter = [energy_converter]

        # Get power distribution
        power_distribution = self._calculate_power_distribution(energy_converter = energy_converter,
                                                                total_power=power_required)

        # Calculate total fuel consumption
        total_consumption = 0.0

        for converter, power in zip(energy_converter, power_distribution):
            if power > 0:
                consumption = converter.fuel_consumption_at_load(
                    load=power,
                    fuel=fuel
                )
                total_consumption += consumption
        self.logger.info(f"Total fuel consumption: {total_consumption}kg of {fuel.fuel_type}")

        return total_consumption * duration_hours
```

#### Calculate energy required function
Calculates the stored energy in GJ required (essential how much stored potential energy you need)
power * time / efficiency (then multiplied by some margin if necessary)

```python
    def calculate_energy_required(self,
                                  power_required: float,
                                  time: float,
                                  converter: EnergyConverter,
                                  fuel: Fuel,
                                  margin: float = 0.0,
                                  fuel_reformer_required: bool = False,
                                  electricity_converter_required: bool = False)-> float:
        """
        Calculate required stored potential energy for the specific power system

        :param power_required: The power that the fuel consumption will be calculated for.
        :type power_required: float
        :param time: The time that the converter is running at the required power level in hours
        :type time: float
        :param converter: EnergyConverter to calculate the required stored energy for
        :type converter: EnergyConverter
        :param fuel: Fuel that the converter is being fueled by.
        :type fuel: Fuel
        :param margin: The margin that will be used for calculating the energy required.
        :type margin: float

        :return: stored energy required in GJ
        :rtype: float
        """
        try:
            system_efficiency = self.calculate_system_efficiency(
                energy_converter=converter,
                power_required=power_required,
                electricity_converter_required = electricity_converter_required,
                fuel_reformer_required = fuel_reformer_required
            )

            # Calculate energy requirement
            base_energy = power_required * time  # kWh
            energy_required = (base_energy * 3.6) / system_efficiency  # Convert to GJ

            # Apply margin if specified
            if margin > 0:
                energy_required *= (1 + margin)

            return energy_required

        except Exception as e:
            self.logger.error(f"Error calculating energy requirement: {str(e)}")
            raise
```

#### Calculate mass and volume of fuel
static methods that calculate the mass and volume of fuel required using the energy required, the LHV and volumetric energy density of the fuel and a safety factor.
static method as they are not specific to the engine room being used.

```python
@staticmethod
    def calculate_fuel_mass(fuel: Fuel,
                            energy_required: float,
                            safety_factor: float = 4) -> float:
        """
        Calculate required fuel mass.

        :param fuel: fuel required.
        :type fuel: Fuel
        :param energy_required: energy required.
        :type energy_required: float
        :param safety_factor: safety factor.
        :type safety_factor: float

        :return: fuel mass in kg
        :rtype: float
        """
        return (safety_factor * fuel.rho * energy_required )/ fuel.cEDV

    @staticmethod
    def calculate_fuel_volume(fuel: Fuel,
                              energy_required: float,
                              safety_factor: float = 4) -> float:
        """
        Calculate required fuel volume.

        :param fuel: fuel required.
        :type fuel: Fuel
        :param energy_required: energy required.
        :type energy_required: float
        :param safety_factor: safety factor.
        :type safety_factor: float

        :return: fuel volume in m^3
        :rtype: float
        """
        return (safety_factor * energy_required)/ fuel.cEDV
```

#### Calculate fuel requirements
Calculates the mass and volume of fuel required for the engine room.
Currently only considers a configuration that is one auxilliary engine, one main engine. 
Doesn't use combination efficiency that was shown above

step 1) energy required for the main engine
step 2) energy required for the aux engine (if needed)
step 3) checks if fuels are the same
        - if they are the same, adds the two stored energies together and calcualtes fuel requirements (mass and voluem)
        -if they are different calcualtes them separately
step 4) returns a dictionary of the different fuel mass and volume.

```python
    def calculate_fuel_requirements(self,
                                    me_power_requirement: float,
                                    me_duration: float,
                                    aux_power_requirement: float,
                                    aux_duration: float,
                                    margin: float = 0.0,
                                    ME_electricity_converter_required: bool = False,
                                    ME_fuel_reformer_required: bool = False,
                                    aux_electricity_converter_required: bool = False,
                                    aux_fuel_reformer_required: bool = False) -> Dict[str, Dict[str, float]]:
        """
        Calculate total fuel requirements for main and auxiliary systems of the specific EngineRoom instance.

        :param me_power_requirement: Power required from the main engine
        :type me_power_requirement: float
        param aux_power_requirement: Power required from the auxiliary engine
        :type aux_power_requirement: float
        param aux_duration: Duration of the auxiliary engine operating at that point
        :type aux_duration: float
        param margin: Margin required for calculating the fuel.
        :type margin: float
        :param ME_electricity_converter_required: Specify whether an electricity converter is required in the main power system
        :type ME_electricity_converter_required: bool
        :param ME_fuel_reformer_required: Specify whether a fuel reformer is required in the main power system
        :type ME_fuel_reformer_required: bool
        :param aux_electricity_converter_required: Specify whether an electricity converter is required in the auxilliary power system
        :type aux_electricity_converter_required: bool
        :param aux_fuel_reformer_required: Specify whether a fuel reformer is required in the auxilliary power system
        :type aux_fuel_reformer_required: bool

        :return:
        :rtype:
        """
        try:
            requirements = {}

            # Calculate main engine requirements
            me_energy = self.calculate_energy_required(
                power_required=me_power_requirement,
                time=me_duration,
                converter=self.ME_converter,
                fuel=self.ME_fuel,
                margin=margin,
                fuel_reformer_required=ME_fuel_reformer_required,
                electricity_converter_required = ME_electricity_converter_required
            )

            # Calculate auxiliary requirements if needed
            aux_energy = 0.0
            if self.aux_converter and aux_power_requirement and aux_duration > 0:
                aux_energy = self.calculate_energy_required(
                    power_required=aux_power_requirement,
                    time=aux_duration,
                    converter=self.aux_converter,
                    fuel=self.aux_fuel,
                    margin=margin,
                    fuel_reformer_required=aux_fuel_reformer_required,
                    electricity_converter_required = aux_electricity_converter_required
                )

            # Calculate fuel masses and volumes
            if self.ME_fuel == self.aux_fuel:
                # Combine requirements for same fuel type
                total_energy = me_energy + aux_energy
                requirements[self.ME_fuel.name] = {
                    'mass (kg)': round(self.calculate_fuel_mass(self.ME_fuel, total_energy), 2),
                    'volume (m3)': round(self.calculate_fuel_volume(self.ME_fuel, total_energy),2)
                }
            else:
                # Separate requirements for different fuels
                requirements[self.ME_fuel.name] = {
                    'mass (kg)': round(self.calculate_fuel_mass(self.ME_fuel, me_energy),2),
                    'volume (m3)': round(self.calculate_fuel_volume(self.ME_fuel, me_energy),2)
                }
                if aux_energy > 0:
                    requirements[self.aux_fuel.name] = {
                        'mass (kg)': round(self.calculate_fuel_mass(self.aux_fuel, aux_energy),2),
                        'volume (m3)': round(self.calculate_fuel_volume(self.aux_fuel, aux_energy),2)
                    }

            return requirements

        except Exception as e:
            self.logger.error(f"Error calculating fuel requirements: {str(e)}")
            raise
```
