### Decomposing engine information into columns

```python
# Example DataFrame with engine information

def extract_engine_models(info_str):
    if not isinstance(info_str, str):
        return []
    # Regex to extract engine model and quantity
    pattern = r'(\d+)_x_([A-Za-z0-9._]+)'
    matches = re.findall(pattern, info_str)
    return matches

# Determine the maximum number of different engines in any row
max_engines = ship_shape['Main_Engine_Summary'].apply(lambda x: len(extract_engine_models(str(x)))).max()

print(f'Maximum number of different engines: {max_engines}')

# Create new columns for each possible engine and quantity pair
for i in range(1, max_engines + 1):
    ship_shape[f'engine_{i}'] = None
    ship_shape[f'quantity_{i}'] = None

for idx, row in ship_shape.iterrows():
    engines = extract_engine_models(str(row['Main_Engine_Summary']))
    total_engines = 0  # Initialize the total engines count
    for i, (quantity, model) in enumerate(engines, start=1):
        ship_shape.at[idx, f'engine_{i}'] = model
        ship_shape.at[idx, f'quantity_{i}'] = int(quantity)
        total_engines += int(quantity)  # Add the quantity to the total count
    ship_shape.at[idx, 'total_engines'] = total_engines  # Store the total count in a new column

print(ship_shape.columns)
```

```python
# List of specific ship types you want to plot
specific_types = ['Bulk_Carrier', 'Cruise_Ship', 'Tanker', 'Ro_Ro/Container']  # Replace with the ship types you are interested in

# Filter the DataFrame for these specific ship types
filtered_data = ship_shape[ship_shape['Type'].isin(specific_types)]

print(filtered_data.columns)
```

```python

# Set the size of the plots
plt.figure(figsize=(12, 8))

# Scatter plot for each ship type
for ship_type in specific_types:
    # Filter data for the current ship type
    group = filtered_data[filtered_data['Type'] == ship_type]
    
    # Create scatter plot
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=group, x='Dwt', y='Main_Engine_1-(mkW)', hue='total_engines', palette='viridis', size='total_engines', sizes=(50, 200))

    plt.title(f'Scatter Plot for Ship Type {ship_type}')
    plt.xlabel('DWT')
    plt.ylabel('Installed  (kW)')
    plt.legend(title='Number of Engines', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True)
    plt.show()
```
