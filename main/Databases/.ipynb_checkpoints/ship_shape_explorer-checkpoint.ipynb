## This notebook is being used to write data handling code

### Importing packages

```python
import numpy as np
import pandas as pd
from dataclasses import dataclass, field
from typing import Optional, List, Dict
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
from sklearn.linear_model import LinearRegression
```

### Importing data

```python
ship_shape = pd.read_excel('ship_data.xlsx')
```

## Focusing on parameters

### Getting names of columns

```python
column_names = ship_shape.columns
print(column_names)

# ship_shape = df['Type','Main Engine Summary' 'GT','Dwt', 'Depth Moulded (m)',
#        'LOA (m)', 'Beam Mld (m)', 'Beam Ext (m)', 'Draft (m)', 'Speed (knots)']

```

### Selecting columns

```python
ship_shape = ship_shape[['Main_Engine_Summary','Type', 'GT', 'Dwt', 'Depth_Moulded_(m)',
       'LOA_(m)', 'Beam_Mld_(m)', 'Beam_Ext_(m)', 'Draft_(m)', 'Speed_(knots)']]

print(ship_shape.head())
```

### removing blanks

```python
specific_columns = ['Dwt', 'LOA_(m)','Beam_Mld_(m)','Draft_(m)','Speed_(knots)']

ship_shape = ship_shape.dropna(subset=specific_columns)
```

### Type mapping the types of ship to a number

```python
# Get a list of all unique values in the 'Type' column
unique_types = ship_shape['Type'].unique()

# Custom keys
custom_keys = {
    'Bulk_Carrier': 1,
    'Tanker': 2,
    'Cruise_Ship': 3,
    'Ro_Ro': 4
}

# Create a priority list with custom keys types at the top
priority_types = list(custom_keys.keys())
remaining_types = [t for t in unique_types if t not in custom_keys]

# Combine the lists
ordered_types = priority_types + remaining_types

# Create a mapping
type_mapping = {type_name: idx + 1 for idx, type_name in enumerate(ordered_types)}

# Create a DataFrame from the mapping dictionary
type_mapping_df = pd.DataFrame(list(type_mapping.items()), columns=['Type', 'Type_Key'])

# Add Type_Key column based on the type_mapping
ship_shape['Type_Key'] = ship_shape['Type'].map(type_mapping)

print(type_mapping_df)
```

#### Creating excel file

```python
with pd.ExcelWriter('ship_data.xlsx', engine='openpyxl') as writer:
    ship_shape.to_excel(writer, sheet_name='Ship Data', index=False)
    type_mapping_df.to_excel(writer, sheet_name='Type Mapping', index=False)
```

## Creating interpolation

```python
def interpolate_beam(ship_type, length):
    # Filter data for the given ship type
    type_data = ship_shape[ship_shape['Type_Key'] == ship_type]
    
    if type_data.empty:
        raise ValueError(f"No data available for ship type: {ship_type}")
    
    # Extract known lengths and beams
    known_lengths = type_data['LOA_(m)'].values
    known_beams = type_data['Beam_Mld_(m)'].values
    
    # Create an interpolation function
    interp_func = interp1d(known_lengths, known_beams, bounds_error=False, fill_value="extrapolate")
    
    # Interpolate the beam for the given length
    return interp_func(length)


L=183
type = 1

interpolate_beam(type,L)
```

## Linear regression

```python
def predict_beam(ship_type, length):
    # Filter data for the given ship type
    type_data = ship_shape[ship_shape['Type_Key'] == ship_type]
    
    if type_data.empty:
        raise ValueError(f"No data available for ship type: {ship_type}")
    
    # Extract known lengths and beams
    known_lengths = type_data[['LOA_(m)']].values
    known_beams = type_data['Beam_Mld_(m)'].values
    
    # Create and fit the linear regression model
    model = LinearRegression()
    model.fit(known_lengths, known_beams)
    
    # Predict the beam for the given length
    length_to_predict = [[length]]
    predicted_beam = model.predict(length_to_predict)[0]
    
    return predicted_beam

L=183
type = 1

print(predict_beam(ship_type = type, length = L))
```

### Grouping per ship type

```python
ship_shape_grouped = ship_shape.groupby('Type')

for type_name, group in ship_shape_grouped:
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=group, x='Dwt', y='LOA_(m)', hue='Type', style='Type', palette='viridis')
    plt.title(f'Scatter Plot for {type_name}')
    plt.xlabel('Dwt')
    
    plt.ylabel('LOA (m)')
    plt.legend(title='Type')
    plt.grid(True)
    plt.show()
```
