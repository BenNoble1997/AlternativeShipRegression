# Clarksons database cleaner
The purpose of this project will be to develop generic tools for converting an excel file downloaded from Clarksons database to a useable file for the digital twin driven design framework. 

#### The initial criteria that was used for the example data
* Bulk Carrier, Cruise Ship, Passenger Vessel, RO-RO, RO-RO Freight, RO-RO/Container RO-RO/LO-LO
* Built after 1997
* Blank cells regarding DWT,GT,LOA,LBP,DRAFT, BEAM, SPEED removed
